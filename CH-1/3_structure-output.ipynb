# Structure output using pydantic models

```python
from pydantic import BaseModel

class llm_schema(BaseModel):
    joke: str
    the_joke: str
```

```python
obj =llm_schema(**{"joke": "Why did the scarecrow win an award?", "the_joke": "Because he was outstanding in his field!","the_joke2": "Because he was outstanding in his field!"})
```

```python
llm_structured_output = llm.with_structured_output(llm_schema)

result = llm_structured_output.invoke("Tell me a joke")
print(result)
```
